```python
# https://dontpad.com/bcc-ds-02

df = pd.read_csv('https://raysonlaroca.github.io/' +
                 'teaching/data-science/datasets/enron.csv')
```

```python
df.head()
```

```python
# método de Tukey

for col in df.select_dtypes(include='number').columns.values:
    print(col)
```

```python
for col in df.select_dtypes(include='number').columns.values:
    print(col)

    # Q1 e Q3
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    # IQR (Q3 - Q1)
    iqr = q3 - q1

    # calcular o limite inferior e o superior (whiskers)
    k = 1.5 # valor "padrão" do método de Tukey (rule of thumb)
    # k = 3 # outliers extremos

    sup = q3 + k * iqr
    inf = q1 - k * iqr

    # obter os dados abaixo ou acima dos limites
    condicao = (df[col] < inf) | (df[col] > sup)
    idx_outliers_col = list(df[condicao].index)

    print(' ', idx_outliers_col, '\n')
```

```python
outliers = []

for col in df.select_dtypes(include='number').columns.values:
    # print(col)

    # Q1 e Q3
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    # IQR (Q3 - Q1)
    iqr = q3 - q1

    # calcular o limite inferior e o superior (whiskers)
    k = 1.5 # valor "padrão" do método de Tukey
    # k = 3 # outliers extremos

    sup = q3 + k * iqr
    inf = q1 - k * iqr

    # obter os dados abaixo ou acima dos limites
    condicao = (df[col] < inf) | (df[col] > sup)
    idx_outliers_col = list(df[condicao].index)

    # print(' ', idx_outliers_col, '\n')

    # armazenar os outliers deste coluna
    for v in idx_outliers_col:
        outliers.append(v)

# lista com repetição dos outliers
print(outliers)
```

# Isolation Forest

```python
from sklearn.ensemble import IsolationForest
```

```python
df.head()
```

```python
df_num = df.select_dtypes(include='number')
df_num
```

```python
# imputação SIMPLES

df_num.fillna(0, inplace=True)
df_num
```

```python
isof = IsolationForest(random_state=123)
isof.fit(df_num)
```

```python
y_pred = isof.predict(df_num)
y_pred

# 1 = valores "normais"
# -1 = outliers
```

```python
df.insert(0, 'outlier', (y_pred == -1).astype(int))
```

```python
df.head()
```

```python
condicao_outlier = df['outlier'] == 1
df[condicao_outlier]
```

```python
scores = isof.score_samples(df_num)
scores
```

```python
df.insert(1, 'outlier_score', isof.score_samples(df_num))
```
