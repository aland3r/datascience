```python
# Seleção condicional
# Exemplo: professores(as) que lecionam a disciplina A

condicao = df['discipline'] == 'A'  # cria uma máscara booleana
df[condicao]  # retorna apenas as linhas que satisfazem a condição
```

```python
# Seleção condicional direta (mesmo resultado)
df[df['discipline'] == 'A']
```

```python
# Seleção condicional composta (E / OU)
# Exemplo: professorAs que lecionam a disciplina A

condicao = df['discipline'] == 'A' and df['sex'] == 'Female'
df[condicao] # erro
```

```python
# As condições precisam estar isoladas entre parênteses,
# mas ainda assim "and" não funciona com Series do pandas
condicao = (df['discipline'] == 'A') and (df['sex'] == 'Female')
df[condicao]  # erro
# Em pandas devemos usar & para "E" e | para "OU"
```

```python
# Operadores lógicos em pandas:
# "and" (Python)  ->  &
# "or"  (Python)  ->  |
# "not" (Python)  ->  ~

condicao = (df['discipline'] == 'A') & (df['sex'] == 'Female')
df[condicao]  # retorna apenas as linhas que satisfazem ambas as condições
```

```python
# Também podemos aplicar a condição diretamente no DataFrame
df[(df['discipline'] == 'A') | (df['sex'] == 'Female')]
```

```python
# Existe diferença no salário médio
# entre professores homens e mulheres?

condicao_m = df['sex'] == 'Male'
condicao_f = df['sex'] == 'Female'

print('Male:  ', df[condicao_m]['salary'].mean())
print('Female:', df[condicao_f]['salary'].mean())
```

```python
# Também podemos calcular isso de outra forma usando GroupBy
# Agrupa os dados pela coluna 'sex' e calcula o salário médio de cada grupo
df.groupby('sex')['salary'].mean()
```

```python
# GroupBy permite analisar os dados em diferentes níveis de granularidade
# Aqui agrupamos por 'sex' e 'rank' e calculamos o salário médio de cada grupo
df.groupby(['sex', 'rank'])['salary'].mean()
```

```python
# Contagem de registros em cada grupo
# Agrupa por 'sex' e 'rank' e conta quantos valores
# de 'salary' existem em cada grupo
df.groupby(['sex', 'rank'])['salary'].count()
```

```python
# E se quisermos mais de uma estatística ao mesmo tempo?
# O método agg permite aplicar várias funções de agregação

df.groupby(['sex', 'rank'])['salary'].agg(['mean', 'median', 'count'])
```

```python
# Aplica uma função a cada valor da coluna 'rank'
# para substituir as abreviações por nomes completos

def tratamento(x):
    if x == 'Prof':
        return 'Professor'
    elif x == 'AsstProf':
        return 'Professor Assistente'
    else:
        return 'Professor Associado'

df['rank'] = df['rank'].apply(tratamento)
```

```python
df.head()
```

```python
# Tratamento usando mais de uma coluna
# Exemplo: True se 'sex' == 'Male' e 'salary' > 150000

def tratamento_duas_colunas(r):
    return r['sex'] == 'Male' and r['salary'] > 150000

# axis=1 faz com que a função seja aplicada linha a linha
df['condicao_duas_colunas'] = df.apply(tratamento_duas_colunas, axis=1)

df.head()
```
